```python
import torch, json
from src.data import get_dataloaders
from src.models import ResNetCNN
from src.train import fit

device = 'cuda'
torch.manual_seed(42)

train_loader, val_loader, test_loader, classes = get_dataloaders(
    root='/kaggle/working/data', batch_size=32, num_workers=0)

model = ResNetCNN(num_classes=37, dropout=0.3)

history = fit(model, train_loader, val_loader,
              epochs=30, lr=1e-3, device=device,
              ckpt_path='models/resnet_cnn_30ep.pth')

with open('reports/resnet_cnn_30ep_history.json', 'w') as f:
    json.dump(history, f)
    
```
